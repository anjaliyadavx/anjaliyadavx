```python
import pandas as pd
data=pd.read_csv('/content/SQL - Retail Sales Analysis_utf .csv')
df=pd.DataFrame(data)
total_sales = df['total_sale'].sum()
print("Total Sales:", total_sales)
category_sales = df.groupby('category')['total_sale'].sum()
print(category_sales)
gender_sales = df.groupby('gender')['total_sale'].sum()
print(gender_sales)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv('/content/SQL - Retail Sales Analysis_utf .csv')
df=pd.DataFrame(data)
category_sales.plot(kind='bar', color=['skyblue', 'lightcoral', 'lightgreen'])
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv('/content/SQL - Retail Sales Analysis_utf .csv')
df=pd.DataFrame(data)
gender_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales by Gender")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv('/content/SQL - Retail Sales Analysis_utf .csv')
df=pd.DataFrame(data)
df['sale_date'] = pd.to_datetime(df['sale_date'], dayfirst=True)
monthly_sales = df.groupby(df['sale_date'].dt.month)['total_sale'].sum()

monthly_sales.plot(kind='line')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
